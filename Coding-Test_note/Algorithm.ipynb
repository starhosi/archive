- 해쉬 (key-value)

```python
# 해쉬값으로 dictionary 만들기
lst = ['A','B']
dic = dict()
for s in lst:
    dic[s] = hash(s)
print(dic)
```

- 힙 (이진트리)

```python
# Heap
import heapq as hq
a = [4,5,2,3,1,7]
print(a)
# lst to heap
print('힙구현')
hq.heapify(a)
print(a)

print('heap pop')
hq.heappop(a)
print(a)

print('heap push')
hq.heappush(a,20)
print(a)
```

```python
# 파이썬은 최소힙만 지원한다. 최대힙은 튜플로 구현
nums = [4, 1, 7, 3, 8, 5]
print('original')
print(nums)

print('최대힙 구현')
heap = []
for num in nums:
  hq.heappush(heap, (-num, num))  # (우선 순위, 값)
print(heap)
a = [x[1] for x in heap]
print(a)

print('max heap push 10')
push_num = 10
hq.heappush(heap,(-push_num, push_num))
print(heap)
a = [x[1] for x in heap]
print(a)

print('max heap pop')
hq.heappop(heap)
print(heap)
a = [x[1] for x in heap]
print(a)
```

- 이진탐색

```python
def binary_search(target, a):
    start, end = 0, len(a)
    mid = (start + end) // 2
    while 0 <= mid < len(a):
        mid = (start + end) // 2
        if target > a[mid]:
            start = mid + 1
        elif target < a[mid]:
            end = mid - 1
        else:
            return mid
    return -1

a = [30,10,2,50,60,2]
a.sort()
print(a)
print(binary_search(30,a))
```

```python
a = [120,110,130,140]
b = 485
# 예산 상한액 구하기 cal이라는 함수로 최소요구값~ 최대요구값 사이의 값들을 이분탐색
def solution(a,b):
    n = int(b/len(a))
    n2 = max(a)

    def cal(n,lst,b):
        rem_b = [i for i in lst if i < n]
        return b - (sum(rem_b)+((len(lst) - len(rem_b))*n))

    if b > sum(a):
        return n2

    start = n
    end = n2
    mid = (start+end)//2
    while end >= start:
        mid = (start+end)//2
        if cal(mid+1,a,b)<0 and cal(mid,a,b) >= 0:
            print(mid)
            break
        if cal(mid,a,b) > 0 :
            start = mid+1
        else:
            end = mid-1
solution(a,b)
```

- 동적계획법(Dynamic Programing)

```python
def fibo(n):
    if n == 1:
        return 1
    if n == 2:
        return 1
    return fibo(n-1)+fibo(n-2)
print(fibo(10))
```

```python
def LCS(i,j):
    x = len(i)
    y = len(j)
    dp1 = [[None] for _ in range(y)]
    dp = [dp1 for _ in range(x)]
    
    for o in range(x):
        for t in range(y):
            if o==0 or t==0:
                dp[o][t] = 0
            elif i[o-1] == j[t-1]:
                dp[o][t] = dp[o-1][t-1]+1
            else:
                dp[o][t] = max(dp[o-1][t],dp[o][t-1])
        
    return dp[x-1][y-1]

LCS('ABCBDAB', 'BDCABA')
```
