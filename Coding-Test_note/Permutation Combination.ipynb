```python
# 순열
from itertools import permutations

data = ['A','B','C','D']

# 4개의 원소를 3개의 원소로 '순서 고려하여' 나열하는 모든 경우의 수
result = list(permutations(data, 3)) 
print(len(result))
print(result)
```

```python
# 순열 중복 허용
from itertools import product

data = ['A','B','C','D']

# 중복 허용한 순열
result = list(product(data, repeat=3)) # repeat 명시 중요
print(len(result))
print(result)
```

```python
# 조합
from itertools import combinations

data = ['E', 'F', 'G']

# 3개의 원소를 2개의 원소로 '순서 고려하지 않고' 나열하는 모든 경우의 수
result = list(combinations(data, 2)) 
print(len(result))
print(result)
```

```python
# 조합 중복 허용
from itertools import combinations_with_replacement

data = ['E', 'F', 'G']

# 3개의 원소를 2개의 원소로 '순서 고려하지 않고' 나열하는 모든 경우의 수
result = list(combinations_with_replacement(data, 2)) 
print(len(result))
print(result)
```
